#### Lista con funciones vendedor

```python
import os

# --- Base de datos temporal ---
vendedores = []

# --- Funciones de Gestión ---

def agregarVendedor():
    print("\n--- AGREGAR VENDEDOR ---")
    id_v = input("ID: ")
    nombre = input("Nombre: ")
    ventas = float(input("Ventas totales: "))

    vendedor = {"id": id_v, "nombre": nombre, "ventas": ventas}
    vendedores.append(vendedor)
    print("✅ Registrado con éxito.")

def listarVendedores():
    print("\n--- LISTADO GENERAL ---")
    if not vendedores:
        print("La lista está vacía.")
    else:
        for v in vendedores:
            print(f"ID: {v['id']} | Nombre: {v['nombre']} | Ventas: ${v['ventas']}")

def buscarVendedor():
    print("\n--- BUSCAR VENDEDOR ---")
    id_buscado = input("Ingrese ID a buscar: ")
    for v in vendedores:
        if v['id'] == id_buscado:
            print(f"🔍 Encontrado -> Nombre: {v['nombre']}, Ventas: ${v['ventas']}")
            return v
    print("❌ No se encontró el vendedor.")
    return None

def modificarVendedor():
    print("\n--- MODIFICAR DATOS ---")
    id_mod = input("ID del vendedor a editar: ")
    for v in vendedores:
        if v['id'] == id_mod:
            v['nombre'] = input(f"Nuevo nombre [{v['nombre']}]: ") or v['nombre']
            v['ventas'] = float(input(f"Nuevas ventas [{v['ventas']}]: ") or v['ventas'])
            print("📝 Datos actualizados.")
            return
    print("❌ Vendedor no encontrado.")

def borrarVendedor():
    print("\n--- ELIMINAR REGISTRO ---")
    id_del = input("ID del vendedor a borrar: ")
    global vendedores
    vendedores = [v for v in vendedores if v['id'] != id_del]
    print("🗑️ Proceso terminado.")

# --- Menú Principal ---

def menu():
    while True:
        print("\n============================")
        print("   SISTEMA DE VENDEDORES   ")
        print("============================")
        print("1. Agregar")
        print("2. Listar")
        print("3. Buscar")
        print("4. Modificar")
        print("5. Borrar")
        print("6. Salir")

        opcion = input("\nSeleccione una opción: ")

        if opcion == '1':
            agregarVendedor()
        elif opcion == '2':
            listarVendedores()
        elif opcion == '3':
            buscarVendedor()
        elif opcion == '4':
            modificarVendedor()
        elif opcion == '5':
            borrarVendedor()
        elif opcion == '6':
            print("👋 Saliendo del sistema...")
            break
        else:
            print("⚠️ Opción no válida.")

# Ejecutar el programa
if __name__ == "__main__":
    menu()
```
